```python
from sentence_transformers import SentenceTransformer

# Load the embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
# ==========================================
# CREATE EMBEDDINGS FOR ALL CHUNKS
# ==========================================

# Extract text from all meeting chunks
texts = [item["chunk"] for item in meeting_chunks]

# Convert all text chunks into embeddings
embeddings = embedding_model.encode(texts)

# Display information
print("Total text chunks:", len(texts))
print("Total embeddings:", len(embeddings))
print("Vector size:", len(embeddings[0]))
```

```python
import faiss
import numpy as np

# Convert embeddings into float32 format
embedding_matrix = np.array(embeddings).astype("float32")

# Get vector dimension
dimension = embedding_matrix.shape[1]

# Create FAISS index
index = faiss.IndexFlatL2(dimension)

# Add all embeddings to the index
index.add(embedding_matrix)

print("FAISS index created successfully!")
print("Total vectors stored:", index.ntotal)
print("Vector dimension:", dimension)
```

```python
# ==========================================
# SEARCH FAISS VECTOR DATABASE
# ==========================================

# User's question
query = "What is Ahmed Khan's budget?"

# Convert the question into an embedding
query_embedding = embedding_model.encode([query])

# Convert to float32 for FAISS
query_embedding = np.array(query_embedding).astype("float32")

# Search for the 3 most similar chunks
k = 3
distances, indices = index.search(query_embedding, k)

print("User Question:")
print(query)

print("\n========== SEARCH RESULTS ==========\n")

for rank, idx in enumerate(indices[0], start=1):
    print(f"Result {rank}")
    print(f"Meeting ID: {meeting_chunks[idx]['meeting_id']}")
    print(f"Text: {meeting_chunks[idx]['chunk']}")
    print(f"Distance: {distances[0][rank-1]}")
    print("-" * 60)
```

```python
# ==========================================
# CREATE A REUSABLE SEARCH FUNCTION
# ==========================================

def search_meeting_notes(query, top_k=3):

    # Convert user question into an embedding
    query_embedding = embedding_model.encode([query])

    # Convert to FAISS-compatible format
    query_embedding = np.array(query_embedding).astype("float32")

    # Search FAISS
    distances, indices = index.search(query_embedding, top_k)

    # Store search results
    results = []

    for rank, idx in enumerate(indices[0], start=1):

        result = {
            "rank": rank,
            "meeting_id": meeting_chunks[idx]["meeting_id"],
            "text": meeting_chunks[idx]["chunk"],
            "distance": float(distances[0][rank - 1])
        }

        results.append(result)

    return results


# ==========================================
# TEST THE FUNCTION
# ==========================================

query = "What is Ahmed Khan's budget?"

results = search_meeting_notes(query)

print("User Question:")
print(query)

print("\n========== SEARCH RESULTS ==========\n")

for result in results:
    print(f"Result {result['rank']}")
    print(f"Meeting ID: {result['meeting_id']}")
    print(f"Text: {result['text']}")
    print(f"Distance: {result['distance']}")
    print("-" * 60)
```

```python
from groq import Groq
import os

client = Groq(api_key=os.environ.get("GROQ_API_KEY"))

print("Groq client ready!")
```

```python
# ==========================================
# FAISS SEARCH + GROQ ANSWER
# ==========================================

def ask_question(question, top_k=3):

    # 1. Convert question into embedding
    question_embedding = embedding_model.encode([question])
    question_embedding = np.array(question_embedding).astype("float32")

    # 2. Search FAISS
    distances, indices = index.search(question_embedding, top_k)

    # 3. Get relevant chunks
    retrieved_chunks = []

    for i in indices[0]:
        retrieved_chunks.append(meeting_chunks[i]["chunk"])

    # 4. Combine retrieved information
    context = "\n\n".join(retrieved_chunks)

    # 5. Send context + question to Groq
    prompt = f"""
You are an assistant that answers questions using the provided meeting information.

Answer the user's question using ONLY the information provided in the context.

If the answer is not present in the context, say:
"I could not find the answer in the provided information."

Context:
{context}

Question:
{question}

Answer:
"""

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0
    )

    answer = response.choices[0].message.content

    print("Question:", question)
    print("\nRetrieved Context:")
    print(context)
    print("\nFinal Answer:")
    print(answer)

    return answer
```

```python
# ==========================================
# FAISS SEARCH + GROQ ANSWER (UPDATED)
# ==========================================

def ask_question(question, top_k=3):

    # Basic error handling
    if not question or not question.strip():
        return "Please enter a valid question."

    # 1. Convert question into embedding
    question_embedding = embedding_model.encode([question])
    question_embedding = np.array(question_embedding).astype("float32")

    # 2. Search FAISS
    distances, indices = index.search(question_embedding, top_k)

    # 3. Get relevant chunks
    retrieved_chunks = []

    for i in indices[0]:
        if 0 <= i < len(meeting_chunks):
            retrieved_chunks.append(meeting_chunks[i]["chunk"])

    # 4. Check if relevant information was found
    if not retrieved_chunks:
        return "I could not find the answer in the provided information."

    # 5. Combine retrieved information
    context = "\n\n".join(retrieved_chunks)

    # 6. Create prompt for Groq
    prompt = f"""
You are an AI assistant for a real estate agency.

Answer the user's question using ONLY the information provided in the context.

Rules:
1. Give the complete answer that directly addresses the question.
2. Include all important details that are relevant to the question.
3. Do not omit relevant information from the context.
4. Do not add information that is not present in the context.
5. If the answer is not present in the context, say:
"I could not find the answer in the provided information."
6. Keep the answer clear and concise.

Context:
{context}

Question:
{question}

Answer:
"""

    # 7. Send request to Groq
    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0
    )

    # 8. Get final answer
    answer = response.choices[0].message.content

    # 9. Display results
    print("Question:", question)
    print("\nRetrieved Context:")
    print(context)
    print("\nFinal Answer:")
    print(answer)

    return answer
```
